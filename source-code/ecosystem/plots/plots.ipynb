The `Plots` package allows to create simple plots easily.

## Requirements

```julia
using Plots
```

## Line plots

```julia
x = 0.0:0.1:10.0
plot(x, sqrt.(x), label="sqrt(x)")
```

It is easy to plot multiple datasets in a single plot.

```julia
x = -10π:0.01:10π;
```

```julia
y1 = @. sin(x)
y2 = @. cos(x);
```

```julia
plot(x, [y1, y2], label=["sin(x)" "cos(x)"])
xlabel!("x")
ylabel!("f(x)")
```

## Scatter plots

```julia
x = 0.0:0.25:10.0
y = sqrt.(x)
y_noise = y + 0.1*randn(length(y));
```

```julia
plot(x, y, label="theory")
```

This plot can be modified by adding a scatter plot of the noisy data.

```julia
scatter!(x, y_noise, label="measurements")
```
